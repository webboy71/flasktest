```python
# @title ComfyUI Model Downloader to Google Drive (via aria2)
import os
from google.colab import drive

# 1. Mount Google Drive
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

# 2. Form Inputs
model_url = "https://civitai.com/api/download/models/3072332?fileId=2951489" # @param {type:"string"}
subfolder = "diffusion_models" # @param ["checkpoints", "diffusion_models", "clip", "text_encoders", "loras", "vae", "unet", "upscale_models"]
custom_filename = "" # @param {type:"string"}
civitai_api_token = "d813cdbc45cf187c4188887ad9f512fa" # @param {type:"string"}
huggingface_token = "" # @param {type:"string"}

# 3. Validate URL and Set Paths
if not model_url.strip():
    raise ValueError("❌ Please paste a valid download URL into the 'model_url' field.")

base_dir = "/content/drive/MyDrive/ComfyUI_gdrive/models"
target_dir = os.path.join(base_dir, subfolder)
os.makedirs(target_dir, exist_ok=True)

# 4. Install aria2 if not already installed
!apt-get -y install -qq aria2 > /dev/null

# 5. Prepare URL & Authentication Headers
final_url = model_url.strip()
header_args = []

if civitai_api_token.strip() and "civitai.com" in final_url:
    separator = "&" if "?" in final_url else "?"
    final_url = f"{final_url}{separator}token={civitai_api_token.strip()}"

if huggingface_token.strip() and "huggingface.co" in final_url:
    header_args.append(f'--header="Authorization: Bearer {huggingface_token.strip()}"')

# 6. Build aria2c Command
# -x 16 -s 16: 16 connections per server for maximum download speed
# -c: Resume partially downloaded files if interrupted
# --content-disposition=true: Automatically detects the real filename from Civitai/HuggingFace headers
filename_arg = f'-o "{custom_filename.strip()}"' if custom_filename.strip() else "--content-disposition=true"
headers_str = " ".join(header_args)

print(f"📂 Target Folder: {target_dir}")
print(f"⬇️ Starting download...\n")

!aria2c -x 16 -s 16 -k 1M -c --file-allocation=none {headers_str} {filename_arg} -d "{target_dir}" "{final_url}"

print(f"\n✅ Done! Saved inside: MyDrive/ComfyUI_gdrive/models/{subfolder}/")
```
